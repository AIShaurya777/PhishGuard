```python
import pandas as pd
import numpy as np
import os
import pandas as pd
import numpy as np
import os
df = pd.read_csv('PhiUSIIL_Phishing_URL_Dataset.csv')
print("🔍 PhiUSIIL Dataset Loaded")
print(f"\nRows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print("📋 Column Names:\n")
for i, column in enumerate(df.columns):
    print(f"{i + 1}. {column}")
print("🔎 First 5 Rows:\n")
display(df.head())
print("📐 Dataset Information:\n")
df.info()
```

```python
print("🎯 Label Distribution:\n")

label_counts = df['label'].value_counts()

print(label_counts)

print("\n📊 Label Percentage:\n")

label_percentage = (
    df['label']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(label_percentage)
print("❓ Missing Values:\n")

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

if len(missing_values) == 0:
    print("No missing values found.")
else:
    print(missing_values)

print(
    f"\nTotal missing values: "
    f"{df.isnull().sum().sum()}"
)
print("❓ Missing Values:\n")

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

if len(missing_values) == 0:
    print("No missing values found.")
else:
    print(missing_values)

print(
    f"\nTotal missing values: "
    f"{df.isnull().sum().sum()}"
)
```

```python
print("=" * 50)
print("🧹 STEP 2: DATA CLEANING")
print("=" * 50)

# Convert URL to string
df['URL'] = df['URL'].astype(str)

# Remove leading and trailing spaces
df['URL'] = df['URL'].str.strip()

# Remove trailing slash
df['URL'] = df['URL'].str.rstrip('/')

print("✅ URL formatting cleaned")
before_duplicates = len(df)

df = df.drop_duplicates(
    subset=['URL']
).reset_index(drop=True)

after_duplicates = len(df)

duplicates_removed = (
    before_duplicates - after_duplicates
)

print(f"Rows before duplicate removal: {before_duplicates}")
print(f"Rows after duplicate removal:  {after_duplicates}")
print(f"Duplicate URLs removed:        {duplicates_removed}")
before_missing = len(df)

df = df.dropna().reset_index(drop=True)

after_missing = len(df)

missing_rows_removed = (
    before_missing - after_missing
)

print(f"Rows before removing missing values: {before_missing}")
print(f"Rows after removing missing values:  {after_missing}")
print(f"Rows removed:                        {missing_rows_removed}")
print("=" * 50)
print("📊 FINAL CLEANED DATASET")
print("=" * 50)

print(f"\nRows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

print(
    f"\nTotal missing values: "
    f"{df.isnull().sum().sum()}"
)

print(
    f"Duplicate URLs: "
    f"{df.duplicated(subset=['URL']).sum()}"
)

print("\n🎯 Label Distribution:")
print(df['label'].value_counts())

print("\n🔎 First 5 Cleaned Rows:")
display(df.head())
CLEANED_FILE = 'phiusiil_cleaned.csv'

df.to_csv(
    CLEANED_FILE,
    index=False
)

print(f"✅ Cleaned dataset saved as: {CLEANED_FILE}")
from google.colab import files

files.download('phiusiil_cleaned.csv')
```
